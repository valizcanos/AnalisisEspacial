### <a href='https://docs.xarray.dev/en/stable/user-guide/terminology.html'> Consulta </a>

```python
import xarray as xr
```

```python
x = xr.DataArray(
    [[25, 35], [10, 24]], #Datos
    dims=("lat", "lon"), #Dimensiones x,y o latitud, longitud
    coords={"lat": [35.0, 40.0], "lon": [100.0, 120.0]}, #Coordenadas x,y o latitud,longitud
)
```

```python
x
```

```python
#Datos diferentes con las mismas dimensiones y coordenadas que 'x'
y = xr.DataArray(
    [[20, 5], [7, 13]],
    dims=("lat", "lon"),
    coords={"lat": [35.0, 42.0], "lon": [100.0, 120.0]},
)
```

```python
y
```

```python
#Alinear
a, b = xr.align(x, y)
```

```python
a
```

```python
b
```

```python
a = xr.DataArray(np.array([1, 2, 3]), dims=["x"])
b = xr.DataArray(np.array([4, 5, 6, 7]), dims=["y"])
```

```python
a
```

```python
b
```

```python
# Crea un arreglo de dos dimensiones cuya forma es (3, 4)
a + b
```

```python
arr1 = xr.DataArray(
    [1, 2, 3], dims=["x"], coords={"x": [10, 20, 30]}, name="arr1"
)
arr2 = xr.DataArray(
    [4, 5, 6], dims=["x"], coords={"x": [20, 30, 40]}, name="arr2"
)
```

```python
arr1
```

```python
unir_datos = xr.Dataset({'Arr1':arr1, 'Arr2':arr2})
unir_datos
```

```python
#Concatenar datos

a = xr.DataArray([[1, 2], [3, 4]], dims=("x", "y"))
b = xr.DataArray([[5, 6], [7, 8]], dims=("x", "y"))
c = xr.concat([a, b], dim="c")
c
```

```python
#Combinar datos
ds1 = xr.Dataset(
    {"data": xr.DataArray([[1, 2], [3, 4]], dims=("x", "y"))},
    coords={"x": [1, 2], "y": [3, 4]},
)
ds2 = xr.Dataset(
    {"data": xr.DataArray([[5, 6], [7, 8]], dims=("x", "y"))},
    coords={"x": [2, 3], "y": [4, 5]},
)

# combine the datasets
combined_ds = xr.combine_by_coords([ds1, ds2], join="outer")
combined_ds
```

```python
airtemps = xr.tutorial.open_dataset("air_temperature")
airtemps
```

```python
# Convertir unidades a Celsius
air = airtemps.air - 273.15

# copiar los atributos a la nueva variable
air.attrs = airtemps.air.attrs
air.attrs["units"] = "°C"
```

```python
air
```

```python
air1d = air.isel(lat=10, lon=10)
air1d.plot();
```
